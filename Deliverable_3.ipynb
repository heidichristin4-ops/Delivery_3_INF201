#### Task 2: Dice class

We implemented:
- An init() method to set the number of sides on the dices.
- A roll() method to generate a random number.
- A epr() method to represent the dice objects.
- A call() method to roll the dice by calling the dice object.


```python
from Dice import Dice

# Create two different dices
d6 = Dice()
d20 = Dice(sides=20)

# Display the dices
print(d6)
print(d20)

# Display Dice attributes
print("D6 sides:", d6.sides)
print("D20 sides:", d20.sides)

# Dice methods
print("\nDice methods:")
print("roll()")
print("__repr__()")
print("__call__()")

# Test roll()
print("Testing roll() ")
print("D6 roll:", d6.roll())
print("D20 roll:", d20.roll())

# Test __call__()
print("Testing __call__() ")
print("D6 roll:", d6())
print("D20 roll:", d20())
```

The tests show that: 
- The default dice has 6 sides.
- We can create dice with different numbers of sides.
- The roll() method returns a random number between 1 and the number of sides.
- The repr() method displays the dice object.
- The call() method allows us to roll the dice by calling the object directly.


#### Task 3: Weapon

We implemented:

- An attack() method that uses weapon damage when equipped.

```python
from Character import Character
from Dice import Dice
from items import Weapon

# Create a weapon
sword = Weapon("Sword", Dice(10))

# Create characters
hero = Character("Hero")
warrior = Character("Warrior", weapon=sword)
enemy = Character("Enemy")

# Display weapon attributes
print("Weapon:", sword.name)
print("Damage dice:", sword.damage_dice)

# Test attack without weapon
hero.attack(enemy)
print("Enemy HP after normal attack:", enemy.hp)

# Test attack with weapon
warrior.attack(enemy)
print("Enemy HP after weapon attack:", enemy.hp)
```

The tests show that the weapon system works as intended.

#### Task 4: Character specialzation

We implemented: 

- A Warrior class with armor to reduce damage. 
- A Mage class with mana for spell attacks.
- An atttack() method for spell and normal attacks.
- Mana regeneration when mana is below 10. 

```python
from Character import Character, Warrior, Mage

# Create characters
warrior = Warrior("Thor")
mage = Mage("Merlin")
enemy = Character("Enemy")

# Test Warrior
print(f"Warrior HP before damage: {warrior.hp}")

# Warrior has armor, so damage should be reduced
warrior.take_damage(20)

print(f"Warrior HP after taking 20 damage: {warrior.hp}")
print(f"Warrior armor: {warrior.armor}")

# Test Mage
print(f"Mage mana before attack: {mage.mana}")
print(f"Enemy HP before attack: {enemy.hp}")

# Mage should cast a spell
mage.attack(enemy)

print(f"Enemy HP after spell attack: {enemy.hp}")
print(f"Mage mana after attack: {mage.mana}")

# Continue attacking until mana is low
for i in range(4):
    mage.attack(enemy)

print(f"Mage mana: {mage.mana}")
```

The tests show that the Warrior and Mage classes work as intended.

- The Warrior's armor reduces incomning damage.
- The Mage uses mana for powerful spell attacks. 
- The Mage regenerates mana when using normal attacks.
